# Signal Denoising 

```python
import numpy as np 
import matplotlib.pyplot as plt
```

## Signal Denoising by moving avarage filter

```python
# Creating signal
srate = 256  # Sampling rate [Hz]
t = np.arange(0, 3, 1/srate )
pnts = len(t)

# Noiseless signal 
signal = np.sin( 2 * np.pi * 2 * t)

# Creating a noise 
noise = 7 * np.random.randn(pnts)

# Adding noise on signal
noisySignal = signal + noise
noisySignal
```

```python
plt.figure(figsize=(25,18))
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.plot(t, noisySignal, 'g', label='Noisy Signal')
plt.title(label = 'Noisy Signal', fontsize = 30)
plt.xlabel('time [s]', fontsize = 20)
plt.ylabel('amplitude', fontsize = 20)
plt.legend(fontsize = 20)

plt.show()
```

```python
# Plotting 

plt.figure(figsize=(25,18))
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.plot(t, filt_sig, 'g', label='Filtered Signal')
plt.title(label = 'Filtered Signal: Moving avarage filter', fontsize = 30)
plt.xlabel('time [s]', fontsize = 20)
plt.ylabel('amplitude', fontsize = 20)
plt.legend(fontsize = 20)

plt.show()
```

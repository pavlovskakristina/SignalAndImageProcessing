# Median Filter

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Creating signal
srate = 256  # Sampling rate [Hz]
t = np.arange(0, 3, 1/srate )
pnts = len(t)
print(pnts)

# Noiseless signal 
signal = np.sin( 2 * np.pi * 2 * t)

# Creating a noise 
noise = 7 * np.random.randn(pnts)

# Adding noise on signal
noisySignal = signal + noise
noisySignal
```

```python
plt.figure(figsize=(25,18))
plt.rcParams['xtick.labelsize'] = 25
plt.rcParams['ytick.labelsize'] = 25
plt.plot(t, noisySignal, 'g', label='Noisy Signal')
plt.title(label = 'Noisy Signal', fontsize = 30)
plt.xlabel('time [s]', fontsize = 30)
plt.ylabel('amplitude', fontsize = 30)
plt.legend(fontsize = 25)

plt.show()
```

```python

N = 50 # ORDER of moving avarage filter

# Initialize denoised signal
filt_Sig = np.zeros(noisySignal.shape[0])
```

```python
# Applying Median filter 

for i in range(0 , noisySignal.shape[0]):
    filt_Sig[i] = np.median(noisySignal[i: N+i])
    print(f"{i}: {filt_Sig[i]} ")
```

```python
# Plotting

plt.figure(figsize=(25,18))
plt.rcParams['xtick.labelsize'] = 25
plt.rcParams['ytick.labelsize'] = 25
plt.plot(t, noisySignal, 'g-', label='Noisy Signal', linewidth= 2.5)
plt.plot(t, filt_Sig, 'r-', label='Median Filtered Signal', linewidth= 2.5)

plt.title(label = 'Noisy Signal VS Median Filtered Signal ', fontsize = 30)
plt.xlabel('time [s]', fontsize = 30)
plt.ylabel('amplitude', fontsize = 30)
plt.legend(fontsize = 25)

plt.show()
```

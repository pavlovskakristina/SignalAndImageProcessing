## Generating Waves

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
```

```python
freq = 256 #hz
time = np.arange(0., 1., 1/freq)

signal1 = 5 * np.sin(2 * np.pi * 2 * time) #amplitude = 5 and frequency = 2 Hz
signal2 = 2 * np.sin(2 * np.pi * 4 * time) #amplitude = 2 and frequency = 4 Hz
signal3 = 7 * np.sin(2 * np.pi * 6 * time) #amplitude = 7 and frequency = 6 Hz

combinedSignal = signal1 + signal2 + signal3 
combinedSignal
```

```python
# Plotting 
plt.figure(figsize=(28,18))
plt.suptitle("Combined Wave (3 waves)", fontsize=30)
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20

plt.subplot(2,2,1)
plt.plot(time, signal1, linewidth=3)
plt.title("amplitude = 5 and frequency = 2 Hz", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)


plt.subplot(2,2,2)
plt.plot(time, signal2, linewidth=3)
plt.title("amplitude = 2 and frequency = 4 Hz", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)

plt.subplot(2,2,3)
plt.plot(time, signal3, linewidth=3)
plt.title("amplitude = 7 and frequency = 6 Hz", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)

plt.subplot(2,2,4)
plt.plot(time, combinedSignal, linewidth=3)
plt.title("Combined Wave", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)

plt.show()
```

## Difference between sinus wave and cosinus wave

```python
freq = 256 #hz
time = np.arange(0., 1., 1/freq)
phi = np.pi/3  #DELETE

SinusWave = 3 * np.sin(2*np.pi*5*time)
CosinusWave = 3 * np.cos(2*np.pi*5*time)

# Plotting 
plt.figure(figsize=(28,18))
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20

plt.plot(time, SinusWave, 'y', linewidth=3, label='Sinus Wave')
plt.plot(time, CosinusWave, 'r', linewidth=3, label='Cosinus Wave')
plt.title("Sinus and Cosinus Wave", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)
plt.xlim([0,1])
plt.legend(fontsize=18)
plt.show()
```

## Constructing a wave with sin, cos, and a DC

```python
freq = 256 #hz
time = np.arange(0., 1., 1/freq)

sinus= np.sin(2 * np.pi * 2 * time) #amplitude = 1 and frequency = 2 Hz
cosinus = 4 *np.cos(2 * np.pi * 4 * time) #amplitude = 2 and frequency = 4 Hz
DC = 2

combinedSign = sinus + cosinus
combinedSignDC = combinedSign + DC
combinedSignDC
```

```python
# Plotting 
plt.figure(figsize=(28,18))
plt.suptitle("Combined Wave with sinus, cosin and DC component", fontsize=30)
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20

plt.subplot(2,2,1)
plt.plot(time, sinus, linewidth=3)
plt.title("SINUS", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)

plt.subplot(2,2,2)
plt.plot(time, cosinus, linewidth=3)
plt.title("COSINUS", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)

plt.subplot(2,2,3)
plt.plot(time, combinedSign, linewidth=3)
plt.title("Combined Signal (sinus + cosinus)", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)

plt.subplot(2,2,4)
plt.plot(time, combinedSignDC, linewidth=3)
plt.title("Combined Signal (sinus + cosinus + DC)", fontsize=25)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('Amplitude', fontsize=20)
plt.show()
```

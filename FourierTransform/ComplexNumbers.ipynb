# Complex Numbers

Comlex number is the sum of real and imaginary number

            z = x + jy

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
```

 ## Different way of creating complex numbers in python

```python
cNum1 = 5 + 6j
cNum2 = 5 + 6* 1j
cNum3 = complex(5, 6)

print(cNum1)
print(cNum2)
print(cNum3)
```

```python
# Extracting real and Imaginary part of complex number
# We use numpy library for that

print(f"Rael number: {np.real(cNum1)}")
print(f"Imaginary number: {np.imag(cNum1)}")
```

```python
## Plotting a complex numbers

cn1 = complex(4, 0)
cn2 = complex(-7, 0)
cn3 = complex(0, -3)
cn4 = complex(0, 5)

# Plotting 
plt.figure(figsize=(28,18))
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20

plt.plot(np.real(cn1), np.imag(cn1), 'yo', markersize= 15, label = cn1)
plt.plot(np.real(cn2), np.imag(cn2), 'mo', markersize= 15, label = cn2)
plt.plot(np.real(cn3), np.imag(cn3), 'bo', markersize= 15, label = cn3)
plt.plot(np.real(cn4), np.imag(cn4), 'ro', markersize= 15, label = cn4)

plt.title("Complex plane", fontsize=25)
plt.xlabel('Real axis', fontsize=20)
plt.ylabel('Imaginary axis', fontsize=20)
plt.xlim([-10,10])
plt.ylim([-10,10])
plt.legend(fontsize=18)
plt.show()
```

```python
## Plotting a complex numbers

cn1 = complex(4, 0)
cn2 = complex(-7, 0)
cn3 = complex(2, -3)
cn4 = complex(2, 3)

# Plotting 
plt.figure(figsize=(28,18))
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20

plt.plot(np.real(cn1), np.imag(cn1), 'yo', markersize= 15, label = cn1)
plt.plot(np.real(cn2), np.imag(cn2), 'mo', markersize= 15, label = cn2)
plt.plot(np.real(cn3), np.imag(cn3), 'bo', markersize= 15, label = cn3)
plt.plot(np.real(cn4), np.imag(cn4), 'ro', markersize= 15, label = cn4)

plt.title("Complex plane", fontsize=25)
plt.xlabel('Real axis', fontsize=20)
plt.ylabel('Imaginary axis', fontsize=20)
plt.xlim([-10,10])
plt.ylim([-10,10])

plt.plot([-10,10], [0, 0], 'g', lw = 3)
plt.plot([0, 0], [-10,10], 'g', lw = 3)
plt.plot([0, 2], [0, 3], 'r-', lw = 3)  # linial function

plt.legend(fontsize=18)
plt.show()
```

## Mathematical operations on complex numbers 

```python
# Addition
cnumb1 = complex(2, 3)
cnumb2 = complex(7, 5)

addition = cnumb1+cnumb2
print(f'addition: {addition}')
```

```python
# Multiplication
multiplication = cnumb1*cnumb2
print(f'multiplication: {multiplication}')
```

```python
# Conjugate of a complex number 
# Basicly just changing + into - 

z1 = complex(4,3)
Conjugated_z1 = np.conj(z1)
print(f'Conjugated number of {z1} is: {Conjugated_z1}')
```

```python
# DIVISION

division = cnumb1/cnumb2
print(f'Devided: {division}')
```

### Magnitude

```python
z = complex(5, 6)

magnitude = np.sqrt(np.real(z)**2 + np.imag(z)**2)
print(f'Magnitude equals: {magnitude}')
```

### Angle with positive real axis

```python
angle = np.angle(z)
print(f'Angle in radians:{angle}')

print(f'Angle in degrees: {angle* 57.3}')
```

```python
# Plotting 

plt.figure(figsize=(28,18))
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20

plt.plot(np.real(z), np.imag(z), 'ko', markersize= 15, label = z)
plt.title("Complex plane", fontsize=25)
plt.xlabel('Real axis', fontsize=20)
plt.ylabel('Imaginary axis', fontsize=20)
plt.xlim([-10,10])
plt.ylim([-10,10])

plt.plot([-10,10], [0, 0], 'k')
plt.plot([0, 0], [-10,10], 'k', lw = 3)
plt.plot([0, 5], [0, 6], 'r-', lw = 3)  # linial function

plt.legend(fontsize=18)
plt.show()
```

# Complex Sinus Wave

```python
t = np.linspace(0, 10, 100)
freq = 50 
A = 1
theta = 0

complex_wave = A * np.exp( -1j * (2 * np.pi * freq * t + theta))
complex_wave
```

```python
# Plotting

plt.figure(figsize=(28,18))
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20

plt.plot(t, np.real(complex_wave), 'b-', linewidth= 3, label = 'real part')
plt.plot(t, np.imag(complex_wave), 'm-', linewidth= 3, label = 'imaginary part')

plt.title("Complex sinus wave", fontsize=25)
plt.xlabel('time', fontsize=20)
plt.ylabel('amplitude', fontsize=20)
plt.xlim([-0.1, 10])

plt.legend(fontsize=15)
plt.show()
```

# Fourier Transform

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generating the signal

srate = 256 #Hz
t = np.arange(0., 1., 1/srate);
amp = 3
f = 5 #Hz
signal = amp * np.sin(2*np.pi * f * t)



# Initializing Fourier Coefficients 
X = np.zeros(len(signal), dtype=complex)   ## 0.+0.j,

for freq in range(0, len(t)):
    #Creating complex sine wave and compute dot product with signal
    csw = np.exp( -1j*2 * np.pi * freq * t )
    X[freq] = np.sum(np.multiply(signal, csw))

# Extract amplitude
amps = np.abs(X) / len(t)
amps
```

```python
# Plotting Signal and Complex Sine Wave

plt.figure(figsize=(24,12))
plt.plot(t, signal, linewidth=3, label="Signal with amplitude = 3 and frequency = 5 Hz")
plt.plot(t, 3*np.real(csw), 'r-', linewidth=3, label="Real part of csw")
plt.plot(t, 3*np.imag(csw), 'm-', linewidth=3, label="Imag part of csw")
plt.xlabel('time', fontsize=15)
plt.ylabel('amplitude', fontsize=15)
plt.legend(fontsize=15)
plt.title("Signal Modeling with complex sine wave", fontsize=25)
plt.xlim([0,1])


plt.show();
```

## Plotting the Signal and it's Fourier Transform

```python
plt.figure(figsize=(24,12))
plt.suptitle("Signal and it's Fourier Transform", fontsize=40)
plt.rcParams['xtick.labelsize'] = 20
plt.rcParams['ytick.labelsize'] = 20


plt.subplot(2, 2, 1)
plt.plot(t, signal, linewidth=3)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('amplitude', fontsize=20)
plt.title("Signal with amplitude = 3 and frequency = 5 Hz")

plt.subplot(2, 2, 2)
markerline, stemlines, baseline = plt.stem(amps)
plt.setp(stemlines, 'linewidth', 3)
plt.xlabel('time [s]', fontsize=20)
plt.ylabel('amplitude', fontsize=20)
plt.title("Fourier Transform of Signal")




plt.show();
```

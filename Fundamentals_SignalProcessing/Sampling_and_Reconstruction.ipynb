# Sampling and Reconstruction

```python
import numpy as np
import matplotlib.pyplot as plt 
from matplotlib import style
```

## Bad Sampling because fs < 2Fmax 

```python
freq = 20 #Hz
t = np.linspace(0, 0.5, 200);
x1 = np.sin(2*np.pi * freq * t)

frequency = 35 # Hz, this frequency is less than the required freq according to Nyguist Sampling Theorem

T = 1/frequency
n = np.arange(0, 0.5/T);
nT = n*T
x2 = np.sin(2*np.pi * freq * nT) # Since for sampling t = nT
```

```python
print(t)
print(nT)
```

```python
# Plotting 

plt.figure(figsize=(25, 20))
plt.suptitle('Sampling a Sinus wave of Fmax = 20Hz with fs = 35Hz', fontsize = 30)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15

plt.subplot(2,2,1)
plt.plot(t, x1, linewidth=3)
plt.title('Sampling a Sinus wave of Fmax = 20Hz with fs = 35Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.subplot(2,2,2)
plt.plot(nT, x2, 'ro', linewidth=3)
plt.title('Sample marks after resampling at fs = 35Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.subplot(2,2,3)
plt.stem(nT, x2, 'm')
plt.title('Sample after resampling at fs = 35Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.subplot(2,2,4)
plt.plot(nT, x2, 'g-', linewidth=3)
plt.title('Reconstructed Sinus Wave fs = 35Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.show()
```

## Good Sampling because fs >= 2Fmax

```python
freq = 20 #Hz
t = np.linspace(0, 0.5, 200);
x1 = np.sin(2*np.pi * freq * t)

frequency = 50 # Hz, this frequency is less than the required freq according to Nyguist Sampling Theorem

T = 1/frequency
n = np.arange(0, 0.5/T);
nT = n*T
x2 = np.sin(2*np.pi * freq * nT) # Since for sampling t = nT
```

```python
# Plotting 

plt.figure(figsize=(25, 20))
plt.suptitle('Sampling a Sinus wave of Fmax = 20Hz with fs = 50Hz', fontsize = 30)
style.use('dark_background')
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15

plt.subplot(2,2,1)
plt.plot(t, x1, linewidth=3)
plt.title('Sampling a Sinus wave of Fmax = 20Hz with fs = 50Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.subplot(2,2,2)
plt.plot(nT, x2, 'ro', linewidth=3)
plt.title('Sample marks after resampling at fs = 50Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.subplot(2,2,3)
plt.stem(nT, x2, 'm')
plt.title('Sample after resampling at fs = 50Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.subplot(2,2,4)
plt.plot(nT, x2, 'g-', linewidth=3)
plt.title('Reconstructed Sinus Wave fs = 50Hz', fontsize = 23)
plt.xlabel('time', fontsize = 23)
plt.ylabel('amplitude', fontsize = 23)

plt.show()
```

Summary: The biggest fs, shows us the most similar reconstructed wave to the fisrt wave (before sampling)

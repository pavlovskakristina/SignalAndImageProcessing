# Discrete Time Signals

```python
# importing libraries 
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
import scipy
from scipy import signal
```

```python
## Generation of unit impulse signal
```

```python
x = []
n = []     # t = nT

for i in range(-5, 6):
    n.append(i)  #time from -5 to 5 
    if i==0:
        x.append(1) # Maximum value
    else:
        x.append(0) 
```

```python
print(x)
print(n)
```

```python

plt.figure(figsize=(20,10))
plt.title('Constructing a Wave', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.stem(n,x, 'yo', label='Unit impulse function') # function
plt.legend(fontsize=20)
plt.ylim(-0.01, 1.2)
plt.xlim(-5, 5)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.show()
```

## Sinusoidal Signal

```python
n = np.linspace(0, 10, 100)
amp = 5 #amplitude 
f = 50 #frequency
x = amp * np.sin(2 * np.pi * f * n)
```

```python
plt.figure(figsize=(21,11))
style.use('dark_background')
plt.title('Sinusoidal Signal', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.stem(n, x, 'yo', label='Sinusoidal Signal') # function
plt.legend(fontsize=15 )
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.xlim(0, 10)
plt.ylim(-5.5, 5.5)
plt.show()
```

## Exponensial Signal

```python
n = np.linspace(0, 10, 100)
amp = 1
x = amp *np.exp(-n)
```

```python
plt.figure(figsize=(21,11))
style.use('dark_background')
plt.title('Exponensial Signal', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.stem(n, x, 'yo', label='Exponensial Signal') # function
plt.legend(fontsize=20)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.xlim(0, 7)
plt.ylim(-0.25, 1.1)
plt.show()
```

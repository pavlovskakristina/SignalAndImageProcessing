# Continuous Time Signals

In this notebook we gonna learn some analog signals like:
- Unit Impulse Signal
- Time shifted impulse:
    - Time Delay
    - Time advance
- Unit step signal (Skok jednostkowy - funkcja Heaviside'a)
- Unit Ramp Signal (Sygnały Narastajace Liniowo)
- Sinusoidal Signal
- Unit Exponential Signal (Jednostkowy sygnał wykadniczy)

```python
# importing libraries 
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
import scipy
from scipy import signal
```

## Generation of unit impulse signal

```python
x = []
t = []

for i in range(-5, 6):
    t.append(i)  #time from -5 to 5 
    if i==0:
        x.append(1) # Maximum value
    else:
        x.append(0) 
```

```python
print(t)
print(x)
```

```python

plt.figure(figsize=(20,10))
plt.title('Constructing a Wave', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.plot(t,x, linewidth = 3, label='Unit impulse function') # function
plt.legend(fontsize=20)
plt.ylim(-0.01, 1)
plt.xlim(-5, 5)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.show()
```

## SciPy - another way of generating impulse 

```python
Impulse = signal.unit_impulse(10, 'mid') # 10 samples of time
```

```python
plt.figure(figsize=(20,10))
plt.title('SciPy -- constructing a Wave', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'green'
plt.rcParams['ytick.color'] = 'green'
plt.plot(np.arange(-5, 5), Impulse, linewidth = 3, label='SciPy: Unit impulse function') # function
plt.legend(fontsize=20)
plt.ylim(-0.01, 1)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.show()
```

# Time Shifted Impulse (Time Delay)

```python
shifted_impulse = signal.unit_impulse(7, 2) # 7 - time samples, 2 - max amplitude
shifted_impulse
```

```python
plt.figure(figsize=(21,11))
plt.title('Time Shifted Impulse (Time Delay)', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.plot(shifted_impulse, linewidth = 3, label='Time Delay') # function
plt.legend(fontsize=20)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.show()
```

# Generation of Unit Step Signal

```python
xampl = []
time = []

for index in range(-5, 10):
    time.append(index)  #time from -5 to 10
    if index>=0:
        xampl.append(1) # Maximum value
    else:
        xampl.append(0) 
```

```python
print(xampl)
print(time)
```

```python
plt.figure(figsize=(21,11))
plt.title('Unit Step Signal', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.plot(time, xampl, linewidth = 3, label='Unit Step Signal') # function
plt.legend(fontsize=20)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.show()
```

# Time Shifted Step (Time Advance)

```python
x = []
t = []

for i in range (-5, 5):
    t.append(i)
    if i >= -2:
        x.append(1)
    else:
        x.append(0) 
```

```python
print(x)
print(t)
```

```python
plt.figure(figsize=(21,11))
plt.title('Time Shifted Step', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.plot(t, x, linewidth = 3, label='Time Shifted Step (Time Advance)') # function
plt.legend(fontsize=20)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.show()
```

# Ramp Signal

```python
x = []
t = []

for i in range(0 ,10):
    t.append(i)
    x.append(i)
```

```python
print(x)
print(t)
```

```python
plt.figure(figsize=(21,11))
style.use('dark_background')
plt.title('Ramp Signal', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.plot(t, x, linewidth = 5, label='Ramp Signal', color='green') # function
plt.legend(fontsize=20)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.xlim(0, len(t)-1)
plt.ylim(0, len(x))
plt.show()
```

# Sinusoidal Signal

```python
t = np.linspace(0, 10, 100)
t
```

```python
amp = 5 #amplitude 
f = 50 #frequency
x = amp * np.sin(2 * np.pi * f * t)
x
```

```python
plt.figure(figsize=(21,11))
style.use('dark_background')
plt.title('Sinusoidal Signal', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.plot(t, x, linewidth = 5, label='Sinusoidal Signal', color='red') # function
plt.legend(fontsize=20)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.xlim(0, 12)
plt.ylim(-5, 6)
plt.show()
```

# Exponensial Signal

```python
t = np.linspace(0, 10, 100)
amp = 1
x = amp *np.exp(-t)
```

```python
plt.figure(figsize=(21,11))
style.use('dark_background')
plt.title('Exponensial Signal', fontsize=35)
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.rcParams['xtick.color'] = 'blue'
plt.rcParams['ytick.color'] = 'blue'
plt.plot(t, x, linewidth = 5, label='Exponensial Signal', color='#008190') # function
plt.legend(fontsize=20)
plt.xlabel('time', fontsize=15)
plt.ylabel('Amplitude', fontsize=15)
plt.xlim(0, 10.5)
plt.ylim(-0.25, 1.25)
plt.show()
```
